```python
# Let's simulate a dataset relevant to the remote work productivity and mental well-being study.
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Simulate a dataset with 100 employees
n = 100
data = pd.DataFrame({
    'Employee_ID': range(1, n+1),
    'Sector': np.random.choice(['IT', 'Education', 'Customer Support'], size=n),
    'Work_Hours_Per_Week': np.random.normal(loc=40, scale=5, size=n).round(1),
    'Productivity_Score': np.random.normal(loc=70, scale=10, size=n).clip(0, 100).round(1),
    'Stress_Level': np.random.randint(1, 6, size=n),  # 1 (Low) to 5 (High)
    'Remote_Experience_Years': np.random.exponential(scale=2, size=n).round(1),
    'Mental_Wellbeing_Score': np.random.normal(loc=65, scale=15, size=n).clip(0, 100).round(1)
})

# Preview the first few rows
data.head()
```
